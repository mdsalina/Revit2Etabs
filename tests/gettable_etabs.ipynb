```python
import comtypes.client

def connect_active_etabs():
    """
    Conecta con una instancia activa de ETABS.
    """ 
    try:
        myEtabsObject = comtypes.client.GetActiveObject("CSI.ETABS.API.ETABSObject")
        SapModel = myEtabsObject.SapModel
        SapModel.SetPresentUnits(8)
        print("Conectado a ETABS")
        return SapModel
    except Exception as e:
        raise ConnectionError("No se pudo conectar a ETABS. Asegúrate de que ETABS esté abierto.") from e

model=connect_active_etabs()
```

```python
fsp=model.DatabaseTables.GetTableForDisplayArray("Slab Property Definitions",GroupName="")
cols,n_row,rows=fsp[2],fsp[3],fsp[4]
col_name=cols.index("Name")
sections_etabs_list=[]
for i in range(n_rows):
    sections_etabs_list.append(rows[i*len(cols)+col_name])

print(sections_etabs_list)        
```
